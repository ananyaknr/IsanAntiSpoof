# IsanAntiSpoof Local Experiment
This notebook runs the `IsanAntiSpoof` experiment locally using the repository files and logs results with MLflow.

## 1. Import Required Libraries
Install dependencies and import Python modules for a local run.

```python
import os
import sys
from pathlib import Path
import subprocess
import pandas as pd
```

## 2. Set Local Repository Path
Point the notebook to the local project root and verify the repository structure.

```python
repo_root = Path('/mnt/data/IsanAntiSpoof') if Path('/mnt/data/IsanAntiSpoof').exists() else Path.cwd()
print('Repository root:', repo_root)
print('Exists:', repo_root.exists())
        
if repo_root.exists():
    print('Top-level files:')
    for path in sorted(repo_root.glob('*')):
        print('-', path.name)
```

## 3. Inspect Project and Data Structure
Confirm the presence of config, data, source, and experiment directories.

```python
if repo_root.exists():
    for folder in ['configs', 'src', 'data', 'experiments', 'checkpoints']:
        print(folder, '->', (repo_root / folder).exists())
```

## 4. Install Local Dependencies
Install Python dependencies from the repository `requirements.txt`.

```python
requirements_path = repo_root / 'requirements.txt'
if requirements_path.exists():
    print('Installing requirements from', requirements_path)
    subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', str(requirements_path)], check=False)
else:
    print('requirements.txt not found at', requirements_path)
```

## 5. Build/Prepare Features
Optionally run protocol generation and feature extraction before training.

```python
print('To build the protocol, run:')
print('python src/data/build_protocol.py')
print('To extract features, run:')
print('python src/features/extract_all.py feature=lfcc')
```

## 6. Train the Model Locally
Run the repository training script with your local configuration and MLflow tracking.

```python
os.chdir(repo_root)
print('Current working directory:', Path.cwd())
mlflow_uri = 'mlruns'
experiment_name = 'isan_antispoof'
command = [
    sys.executable, 'src/training/train.py',
    'experiment=e1_baseline',
    f'mlflow.tracking_uri={mlflow_uri}',
    f'mlflow.experiment_name={experiment_name}'
]
print('Running:', ' '.join(command))
subprocess.run(command, check=False)
```

## 7. Evaluate Local Results
Review the result CSV and checkpoint artifacts produced by the local run.

```python
results_path = repo_root / 'experiments' / 'results.csv'
print('Results path:', results_path)
if results_path.exists():
    display(pd.read_csv(results_path).tail(10))
else:
    print('No results file found yet.')
```

## 8. Save Artifacts Locally
Inspect saved checkpoints and optionally compress results for sharing.

```python
ckpt_dir = repo_root / 'checkpoints'
print('Checkpoint directory exists:', ckpt_dir.exists())
if ckpt_dir.exists():
    for path in sorted(ckpt_dir.glob('**/*'))[:50]:
        print(path.relative_to(repo_root))
else:
    print('No checkpoints found.')
```
